```python
#Create a numpy array of size 2 x 2

import numpy as np

array_2x2 = np.array ([
    [1,2],
    [3,4]
])

print("This is my numpy 2x2 array")
print(array_2x2)
```

```python
#Create a random numpy array of size 3 x 4

random3x4 = np.random.rand(3,4)

print("Random 3x4 array")
print(random3x4)
```

```python
#Create a numpy array of ones and a numpy array of zeroes

ones_array = np.ones((3,3))
zeros_array = np.zeros((3,3))

print("Array of Ones")
print(ones_array)

print("Array of Zeros")
print(zeros_array)
```

```python
#Create a random numpy array of size 6x3 and transpose it

random6x3 = np.random.rand(6,3)

transposed_3x6 = random6x3.T

print("Original 6x3 array")
print(random6x3)

print("Transposed 3x6 array")
print(transposed_3x6)
```
